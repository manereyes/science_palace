## XX - Normal Distribution

Using SciPy to generate normal distributed values randomly and to compute percentages over the distribution.

```python
from scipy.stats import norm
from matplotlib import pyplot as plt
import numpy as np
```

### Generating random numbers for a normal distribution

With norm.rvs(mean, std, size)

```python
# Metrics
mu = 174
std = 7

# Generating 2000 random values (heights)
r_heights = norm.rvs(mu, std, size=2000)

# Getting the mean and stdv
# mu, std = norm.fit(r_heights)
```

```python
# Plotting the histogram of the heights
plt.hist(r_heights, bins=10, density=True, alpha=0.6, color='g')

# Creating X range and Y curve
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p = norm.pdf(x, mu, std)

# Overlaying the normal dist curve
plt.plot(x, p, 'k', linewidth=2, color='red')
plt.legend()
plt.show()

plt.show()
```

### Percentage of heights shorter than 170

using norm.cdf(value, mean, std)

```python
# Calculating the area of heights shorter than 170
shorter_170 =norm.cdf(178, mu, std)

print(f"Percentage of heights shorter than 170: {shorter_170*100:.2f}%")
```

### Percentage of heights larger than 170

Using norm.cdf()

```python
# Calculating the area of heights larger than 170
larger_170 = 1 - norm.cdf(178, mu, std)

print(f"Percentage of heights larger than 170: {larger_170*100:.2f}%")
```

### What percent of heights are between 170 - 180

```python
# Calculating the area of heights larger than 170
between_180_180 = norm.cdf(180, mu, std) - norm.cdf(170, mu, std)

print(f"Percentage of heights between 170 and 180: {between_180_180*100:.2f}%")
```

### What height are 60% of the observations shorter than? (percentiles)

with norm.ppf()

```python
height_of_60_short = norm.ppf(0.6, mu, std)

print(f"Height of 60% shorter than: {round(height_of_60_short, 2)} cm")
```

### What height are 60% of the observations higher than? (percentiles)

```python
height_of_60_large = norm.ppf((1-0.6), mu, std)

print(f"Height of 60% shorter than: {round(height_of_60_large, 2)} cm")
```
