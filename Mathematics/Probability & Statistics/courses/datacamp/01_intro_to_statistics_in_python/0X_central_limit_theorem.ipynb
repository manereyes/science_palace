## Libraries

```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
```

## Rolling a dice manually X number of times

```python
die = pd.Series([1, 2, 3, 4, 5, 6])
```

```python
# rolling the dice 5 times
roll_5_times = die.sample(5, replace=True)
print(list(roll_5_times))
```

We get different results each time we "roll" the dice X number of times

```python
# Taking the mean of the roll
print(np.mean(roll_5_times))
```

## Rolling a dice 5 times 10 times (loop)

```python
loop_times = 10
dice_times = 5
```

```python
# List to store the means
sample_means = []
```

```python
for i in range(loop_times):
    sample_rolls = die.sample(dice_times, replace=True)
    sample_means.append(np.mean(sample_rolls))
```

```python
print(sample_means)
```

### Visualizing the sample mean distribution

```python
plt.hist(sample_means, bins=10)
plt.show()
```

### Now, 100 sample means

```python
loop_times = 100

sample_means = []

for i in range(loop_times):
    sample_rolls = die.sample(dice_times, replace=True)
    sample_means.append(np.mean(sample_rolls))
```

```python
plt.hist(sample_means, bins=10)
plt.show()
```

The sample means distribution starts to follow a normal distribution, despite the dice rolls distribution is uniform

### Now, 1000 samples

```python
loop_times = 1000

sample_means = []

for i in range(loop_times):
    sample_rolls = die.sample(dice_times, replace=True)
    sample_means.append(np.mean(sample_rolls))
```

```python
plt.hist(sample_means, bins=10)
plt.show()
```

This is the **Central Limit Theorem**

The sampling distribution of a statistic becomes closer to the normal distribution as the number of trials increases.

For this to happen, samples must be:
- Random
- Independent

### CLT on Standard Deviation

```python
# Calculating the standard deviation of 5 dice rolls rolled 1000 times
sample_stds = []

for i in range(1000):
    sample_stds.append(np.std(die.sample(5, replace=True)))
```

```python
# Visualizing
plt.hist(sample_stds, bins=10)
plt.show()
```
