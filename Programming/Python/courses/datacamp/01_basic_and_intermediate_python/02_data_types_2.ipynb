## STR methods

To use a method, simply call it with a "." (dot) next to your variable. Methods often change and are different depending of the data type.

```python
# .replace() method
# Replaces a string for another

favorite_car = "My favorite car model is Mercedes-Benz"
print(favorite_car)
```

```python
favorite_car = favorite_car.replace("Mercedes-Benz", "Audi")
```

```python
print(favorite_car)
```

```python
# .lower() & .upper() methods
# Converts to lowercase or uppercase

alphabet = "ABCDEF...XYZ"
```

```python
alphabet = alphabet.lower()
```

```python
print(alphabet)
```

```python
alphabet = alphabet.upper()   #back to normal
```

```python
print(alphabet)
```
