# Lambda Functions

A _"lambda function"_ is a common custom function that can be reduced to a "one line function".

A tipical custom function consists of...

```python
def get_average(values):               # Defined function
    avg = round((sum(values) / len(values)), 1)    # Body
    return avg                         # Output
```

A lambda funcion consists of : _lambda arguments: expression_

```python
get_average_lambda = lambda x: round((sum(x) / len(x)), 1)

# "lambda" = custom keyword
# "x" = the argument
# "expression" = the code to execute like in a custom variable 
```

Let's test...

```python
grades = [8.0, 9.6, 8.7, 6.3, 9.6, 10.0, 7.5, 8.0, 9.3]
```

```python
print(get_average(grades))
```

```python
print(get_average_lambda(grades))
```

Calling a lambda function without variables/arguments.

```python
print((lambda x: round((sum(x) / len(x)), 1))(grades))    # We use "(lambda)(arguments)"
```

Lambda function with two arguments.

```python
sum = lambda x, y: x+y
```

```python
print(sum(2, 2))
```

## Lambda Functions with Iterables

"map()" is a function that can apply a function to all elements in an iterable...

For a list of prices called "prices", we need to apply a 25% discount on every price.

```python
prices = [599.99, 255.50, 300.00, 210.10]
```

Instead of doing...

```python
prices_w_discount = []

for price in prices:
    new_price = round((price - (price*0.25)), 2)
    prices_w_discount.append(new_price)
```

```python
print(prices_w_discount)
```

We can just...

```python
discount = map(lambda x: round((x - (x*0.25)), 2), prices)
```

```python
print(discount)   # The "map" function returns a map object, we have to cast it as a list
```

```python
print(list(discount))   # magic
```
