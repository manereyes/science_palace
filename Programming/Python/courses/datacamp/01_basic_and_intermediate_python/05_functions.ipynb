# Functions

A function is a piece of code that can be run everytime when it's called to execute repetitive code.

Instead of having this code...

```python
grades = [8.0, 9.6, 8.7, 6.3, 9.6, 10.0, 7.5, 8.0, 9.3]

avg_grades = sum(grades) / len(grades)
round_avg_grades = round(avg_grades, 1)

print(round_avg_grades)
```

You can create a function to do the same code.

```python
def get_average(values):  # get_average function requires "values" argument
    avg = sum(values) / len(values)
    round_avg = round(avg, 1)
    return round_avg  # returns float
```

To use a function stored inside the code, you can call it inside a variable.

```python
avg_grades = get_average(grades)
```

```python
print(avg_grades)
```

Or just print it...

```python
print(get_average(grades))
```

**What is an "argument"?**

These are values provided to a function or a method. They can be...
- Positional Arguments : Provided in order, separated by commas.
- Keyword Arguments : Provide arguments by assigning values to keywords.
- Default Arguments : A default argument that is optional for the user to modify, but essential for the function to work.
- Arbitrary Positional Arguments : Accepts any number of arguments with "*args" or any other name.
- Arbitrary Keyword Arguments : Accepts any number of positional arguments with "**kwargs" or any name

```python
# For example

a_number = 0
a_list = []

def some_function(number, something, flag=False):   # This function does nothing, just for example use
    pass


func_var = some_function(
    a_number,  # positional argument
    something=a_list # Keyword argument
)
```

Wrapping up everything we can modify the "average" function.

```python
def get_average(values, rounded=False):
    avg = sum(values) / len(values)
    if rounded == True:
        round_avg = round(avg, 1)
        return round_avg  # returns float
    else:
        return avg
```

```python
print(get_average(grades))  # without "rounded" argument
```

```python
print(get_average(grades, rounded=True))  # with "rounded" argument
```

Now, using Arbitrary Positional Arguments.

```python
grades = [8.0, 9.6, 8.7, 6.3, 9.6, 10.0, 7.5, 8.0, 9.3]
```

```python
# Creating the average function with Arbitrary Positional Arguments

def get_average(*numbers):
    return round((sum(numbers) / len(numbers)), 1)
```

```python
print(get_average(8.0, 9.6, 8.7, 6.3, 9.6, 10.0, 7.5, 8.0, 9.3))  # Returns the same result, without giving a list
```

```python
# The "*" can create a single iterable

print(get_average(*[8.0, 9.6, 8.7], *[6.3, 9.6, 10.0], *[7.5, 8.0, 9.3]))
```

```python
print(get_average(grades))  # But we will get an error if we provide a list as an argument for an Arbitrary Positional Arg
```

Now, Arbitrary Keyword Arguments.

```python
def get_average(**numbers):
    return round((sum(numbers.values()) / len(numbers.values())), 1)   # We use kwargs.values()
```

```python
print(get_average(a=8.0, b=9.6, c=8.7, d=6.3, e=9.6, f=10.0, g=7.5, h=8.0, i=9.3))  # Unlike *args, we have to declare each argument as a variable=value
```

**kwargs sets every variable and value as a dictionary.

```python
# Thanks to that, we can call the function with a unique argument as a dictionary like this... 

print(get_average(**{"a":8.0, "b":9.6, "c":8.7, "d":6.3, "e":9.6, "f":10.0, "g":7.5, "h":8.0, "i":9.3}))
```

OR we can create a single iterable...

```python
print(get_average(**{"a":8.0, "b":9.6, "c":8.7}, **{"d":6.3, "e":9.6, "f":10.0}, **{"g":7.5, "h":8.0, "i":9.3}))
```

**When to make a custon function?**

- Reduce code complexity.
- When a task needs to be executed with a high frequency. (Don't Repeat Yourself)
- To reduce the number of line.

## Docstrings

A good practice is to place Docstrings at the beginning of a function that explain how the function works.

```python
def get_average(values, rounded=False):
    """Custom function that rounds a list of student grades."""
    avg = sum(values) / len(values)
    if rounded == True:
        round_avg = round(avg, 1)
        return round_avg  # returns float
    else:
        return avg
```

```python
# Calling a docstring of a function
print(get_average.__doc__)
```

One-line docstrings are OK if your function is simple, there are many different ways of documenting your functions with docstrings. The best practice to document a function is...

```python
def get_average(values, rounded=False):
    """
    Custom function that rounds a list of student grades.
    
    Args:
        values (list): A list of numeric values (int/float).
        rounded (bool): A bool to return the rounded result.
        
    Returns:
        round_avg (float): If "True" in rounded (bool). Returns the mean of values list, rounded to a single decimal.
        avg: If "False in rounded (bool) (default value), Returns the mean of values list.
    """
    avg = sum(values) / len(values)
    if rounded == True:
        round_avg = round(avg, 1)
        return round_avg  # returns float
    else:
        return avg
```

```python
print(get_average.__doc__)
```

## Extra : Built-in Functions in Python

PYthon comes with many built-in functions to call.

### min() and max()

Returns the minimum and maximum of a list.

```python
numbers = [0, 45, 12, 58, 9935, 3248, 2343, 96969696, 120489102, 52]
```

```python
print(min(numbers))
print(max(numbers))
```

### sum() and round()

Returns the sum of numbers of a list and the rounded number of a float.

```python
numbers = [2, 34, 5.67]
```

```python
result = sum(numbers)
```

```python
print(result)
print(round(result, 1))  # The first argument is the number to round, the second is the number of decimals to round
```

### len()

Returns an integer, which is the lenght of a list, dictionary, string, etc...

```python
word = "Python is easy"
```

```python
print(f"The word has {len(word)} words...")
```
