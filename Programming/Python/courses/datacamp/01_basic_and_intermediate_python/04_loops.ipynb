# Loops

A "loop" is a piece of code that repeats itself recursively until a condition is met.

Loops are useful when you need to execute a repetitive task.

## "For" loop

You can read it as:

"_for_ each _value_ in _sequence_:
perform this _action_"

Where:
- "_sequence_" is an iterable (a range, a list, a dictionary, a tuple, etc...)
- "_value_" is an iterator (index)

```python
weekdays = ["Monday","Tuesday", "Wednesday", "Thursday", "Friday"]
```

```python
for day in weekdays:
    print(day)
```

You can loop through strings.

```python
word = "awesome"

for letter in word:
    print(letter)
```

Looping through dictionaries.

```python
game = {
    "id": 1,
    "name": "Assassin's Creed",
    "company": "Ubisoft",
    "genre": "Action"
}
```

```python
for key, value in game.items():   # Always use "item" method to pack the key-values into tuples to iterate
    print(f"Key: {key}, Value: {value}")
```

We can alter/update data at the same time we're iterating it, and even set conditional statements.

```python
# See if every age on the list will be able to watch the "Terrifier 3" movie after 5 years
ages_list = [12, 14, 4, 16]

for age in ages_list:
    updated_age = age + 5
    if updated_age >= 18:
        print("You can watch the Terrifier 3 movie")
    else:
        print(":(")
```

Using "Range" function to iterate.

Range function works as "range(start,end+1)"

"end+1" happens because the last index is exclusive.

```python
for number in range(1,11):  # Prints numbers from 1 to 10
    print(number)
```

Handling "List" iterations with indexes and values.

```python
# Printing their indexes
for idx, day in enumerate(weekdays):   #Use "enumerate" to pack every value inside the list with their index
    print(f"Day {idx} is {day}")
```

```python
# Another way to get the same result is with range() function and 
for i in range(len(weekdays)):
    print(f"Day {i} is {weekdays[i]}")  # A smart move is to make a range from 0 to the lenght of the list, and then iterating from the same list with the value of "i"
```

## "While" loop

A While loop executes itself until a condition is different from "True" statement, but not specifically has to be a "true" boolean variable.

Unlike an "If" statement, a While loop performs an action _n_ amount of times until a condition is different from True.

An "If" statement, only performs an action if a condition is True

```python
# ATM While loop

is_empty = False
total_money = 5000

while is_empty != True:  # While is_empty is different from True, this snippet will continue executing, because the condition statement will continue to be True.
    total_money -= 1000
    print(f"A new customer has taken $1000 from the ATM. Total money left: {total_money}")
    
    if total_money <= 0:  # If statement that changes the boolean to "True" once money runs out.
        is_empty = True

print("The ATM needs refilling...")
```

```python
# Simple RPG combat While loop with random hits (using random library)
# This While will execute itself "n" amount of times depending of the random generated numbers.

import random

total_enemy_hp = 100
hit_count = 0   # Adding a "hit" counter

print(f"The boss has {total_enemy_hp} HP. The battle begins...")

while total_enemy_hp >= 0:   # This While loop will continue executing until the total hp falls to 0 or less
    hit_count += 1 # Incresing the number of hits
    random_hit = random.randint(1,50)  # Generating a random hit
    total_enemy_hp -= random_hit  # substracting the hit from the total HP
    print(f"Hit number {hit_count}: The boss takes {random_hit} damage! ... Total health: {total_enemy_hp}")  # Printing the status of the hit and current health
    
print("THE BOSS HAS BEEN DEFEATED!")
```

# The "IN" and "NOT" Keyword

Along the prevous statements and loops, we have used arithmetic and logic operators.

"in" checks if a value is IN a variable/data structure

```python
# Could be strings...

print("s" in "Pizza")  # Always returns a Boolean
```

```python
# Could be lists...

country_list = ["Canada", "México", "USA"]

print("México" in country_list)
print("Germany" not in country_list)
```

```python
# Or entire dictionaries; Keys or values.

currency = {
    "MXN" : 460,
    "USD" : 200,
    "CAD" : 724
}
```

```python
print("MXN" in currency.keys())
print("USD" not in currency.keys())
```
