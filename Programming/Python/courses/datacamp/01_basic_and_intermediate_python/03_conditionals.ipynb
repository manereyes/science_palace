# Conditionals

Conditionals work with _"Boolean"_ data to make comparisons between statements.

## Comparison Operators

Conditions need _"Numeric Comparison Operators"_ or _"Logic Comparison Operators"_ to work, which can be symbols or combinations or symbols to compare values.

Numeric:
- "==" equals / is
- "!=" is not equal / is not
- ">" greater/more than
- "<" less than
- ">=" greater/more or equal to
- "<=" less or equal to


Logic:
- "and"
- "or"

```python
# A little glance of the operators and equality

a = ("You" != "Me")  # Returns "True"
b = ("Me" == "mE")  # Returns "False"
c = (True == True)  # Returns "True"
x = (1 == 1)  # Returns "True"
y = (2 > 3)  # Returns "False"
z = (5 >= 5)  # Returns "True"
```

## Building a Conditional Statement.

"if" condition is met, it executes code inside their snippet-

```python
age = 14

if age < 18:
    print("An underage is not able to drive")
    
# In this case, the condition is met, so it executes the "print" code
```

"elif" statement can be used if you want to provide an extra condition to execute if the first doesn't trigger

```python
age = 20

if age < 18:
    print("An underage is not able to drive")
elif age > 18:
    print("You're abe to drive")
    
# In this case, the second condition is met, so it executes the "print" code
```

Lastly, an "else" condition can be used to execute a last statement where no conditio is met.

For this case, I will add a range of ages using "and".

```python
age = 50 # Change this value as much as you want to test

if age >= 70:
    print("You should not be driving due to your age")
elif age >= 50 and age < 69:
    print("You need a special permission to drive")
elif age >= 18 and age < 50: 
    print("You can have a permission to drive")
else:
    print("An underage is not allowed to drive")
```
