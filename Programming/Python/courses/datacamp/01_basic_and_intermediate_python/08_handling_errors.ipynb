# Error Handling

An error will cause a program to terminate.

Common errors:
- TypeError : Incompatible Data Type.

```python
print("Something" + 6)
```

- ValueError : The value is not within an acceptable range.

```python
int("something")
```

**What is a "Traceback"?**

a Traceback is a report which tells you what type of error occurred.

## How to handle an error in our code?

It is a good practice to ancitipate how errors might occur, specially if someone else could run your code.

Lets build a custom function which require a specific data type to run.

```python
def twentyfive_discount_off(number):
    return round((number - (number*0.25)), 2)
```

If we call the function with a "string" argument, the code will trigger a TypeError

```python
print(twentyfive_discount_off("something"))
```

We can prevent thiw with the "try" and "except".
- Try : executes code, the body of the function.
- Excepts : Tells the function what to execute if the function results in an error.

```python
def twentyfive_discount_off(number):
    try:
        return round((number - (number*0.25)), 2)
    except:
        return "Please provide a correct data type (int/float)"
```

```python
print(twentyfive_discount_off("something"))
```

In other cases, "raise" will be useful. As the name says, "raise" will raise an error on the terminal, works along with "If" statements.

```python
# Lets rebuild the previous function with the "rise" approach

def twentyfive_discount_off(number):
    if type(number) in (int, float):
        return round((number - (number*0.25)), 2)
    else:
        raise TypeError("Please provide a correct data type (int/float)")
```

```python
print(twentyfive_discount_off("something"))
```

The _TypeError: can't multiply sequence by non-int of type 'float'_ was replaced with our custom raise error.

**If both approaches works the same... What's the difference?**

- Try-Except will avoid errors, but the program will not be terminated, subsequent code will be still executed.
- Raise terminates code, this is useful for critical functions that will worse the code functionality if something goes wrong.
