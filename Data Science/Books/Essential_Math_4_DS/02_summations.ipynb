# Summations

Python can be useful for _summations_. Considering a summation like the following...

$$\sum_{i=0}^{5} 2i = (2)1 + (2)2 + (2)3 + (2)4 + (2)5 = 30$$

We can code it in python like:

```python
# The old-fashioned way to do a summation
# for loop and storing results in a list.

results_list = []

for n in range(1,6):
    results_list.append(2*n)
```

```python
summation = sum(results_list)
```

```python
print(summation)
```

```python
# With list-comprehension

summation = sum(2*n for n in range(1,6))
```

```python
print(summation)
```

When _"n"_ represent a number of items in a collection, like the number of records in a dataset, it can be presented in python as a list.

Another example with iteration of number collections of the size of _n_.

$$\sum_{i=1}^{n} 10x_{i} = 130$$

In maths, the _n_ value is often started at _1_, unlike in programming (which starts at 0)

```python
x = [1, 4, 6, 2] #  collection of numbers that will be multiplied by "10"
n = len(x)  # the number of items in the collection
```

```python
summation = sum(10*x[i] for i in range (0, n))
```

```python
print(summation)
```

# Summations in SymPy

$$\sum_{i=0}^{5} 2i = 30$$

```python
from sympy import *
```

```python
i, n = symbols('i n') # declaring "i" and "n" variables
```

```python
summation = Sum(2*i, (i, 1, n)) # iterating each element from 1 to n, thhen multiplying them and sum
```

```python
up_to_5 = summation.subs(n, 5) # specifying n as 5, then, iterating 1 through 5
```

```python
print(up_to_5.doit()) # printing the executed expression
```
