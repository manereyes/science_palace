# Linear and Curvilinear Functions

```python
from sympy import *
```

_This is known as a Linear Function._ 

$$f(x) = 2x + 1$$

```python
# Declaring 'x' and the f(x) linear function
x = symbols('x')
f = 2*x + 1
```

```python
# Plotting a cartesian plane
plot(f)
```

_This is known as a "parabola"._

$$f(x) = x^2 + 1$$

```python
x = symbols('x')
f = x**2 + 1
```

```python
plot(f)
```

_Also, functions that produce a continuous but curvy, rather than a linear and straight line, are called "curvilinear function"._

# Three Dimentional Chart

_Functions with two independent variables (x and y) and one dependent variable (the output of f(x, y)) need a three dimentions graph._

```python
from sympy import *
from sympy.plotting import plot3d
```

$$f(x,y) = 2x + 3y$$

```python
x, y = symbols ('x y')
f = 2*x + 3*y
```

```python
plot3d(f)
```
